```python
import pandas as pd

dane = pd.read_csv('../data/ab_data.csv')
country = pd.read_csv('../data/countries.csv')
```

```python
fulldata = dane.merge(country, on = 'user_id', how='left')
```

```python
correcteddata =fulldata[((fulldata["group"]=="control") & (fulldata["landing_page"]=="old_page")) |((fulldata["group"]=="treatment") & (fulldata["landing_page"]=="new_page"))]
```

```python
control_cov = correcteddata[((correcteddata['group']=='control') & (correcteddata['converted']==1))]
```

```python
treatment_all = correcteddata[(correcteddata['group']=='treatment')]
treatment_cov = correcteddata[((correcteddata['group']=='treatment') & (correcteddata['converted']==1))]
```

```python
lista_krajow = list(correcteddata['country'].unique())
```

```python
import statsmodels.api as sm

for c in lista_krajow:
    control_all = correcteddata[(correcteddata['country'] == c) & (correcteddata['group'] =='control') ]
    control_cov = correcteddata[(correcteddata['country'] == c) & (correcteddata['group'] =='control') & (correcteddata['converted']==1)]
    treatment_all = correcteddata[(correcteddata['country'] == c) & (correcteddata['group'] =='treatment') ]
    treatment_cov = correcteddata[(correcteddata['country'] == c) & (correcteddata['group'] =='treatment') & (correcteddata['converted']==1)]

    sukces = [len(control_cov), len(treatment_cov)]
    trials = [len(control_all), len(treatment_all)]

    Z, p = sm.stats.proportions_ztest(sukces, trials)
    print("Country: ", c)
    print("Z-stat:", Z)
    print("p-value:", p)
    
```
